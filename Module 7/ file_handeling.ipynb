```python
with open("./sample_data/sample.csv", "r") as file:
  for line in file:
    print(line)
```

**Get the text from sample.txt , and count the following**
**Number of lines**
**Number of words**
**Number of characters**
**and save it to the counter_of_string.txt file**

```python
with open("./sample_data/test.txt", "r") as file:
  string = file.readlines()
  total_line = len(string)
  print("Total line:",total_line)

  number_of_word = list(map(lambda x: len(x.split()),string))
  print("Total word:",sum(number_of_word))


  string_list = list(map(str.strip, string))
  # print(string_list)
  string_list_space = list(map(lambda x: x.replace(" ", ""), string_list))
  print(string_list_space)

  number_of_char = list(map(lambda x: len(x),string_list_space))
  print("Total Characters:",sum(number_of_char))


with open("./sample_data/counter_of_string.txt","w") as file:
  file.write(f"total line: {total_line}\ntotal number of words:{sum(number_of_word)}\ntotal number of char: {sum(number_of_char)}")


```
