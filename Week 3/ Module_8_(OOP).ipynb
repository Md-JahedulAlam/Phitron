```python
class Phone:
  category = "Electronics"

  # Constructor
  def __init__(self, model, battery, camera):
    self.model = model
    self.battery = battery
    self.camera = camera
```

```python
iphone = Phone("Iphone 17", 4000, 40)
print(iphone.model)
print(iphone.battery)
print(iphone.camera)
print(iphone.category)
```
