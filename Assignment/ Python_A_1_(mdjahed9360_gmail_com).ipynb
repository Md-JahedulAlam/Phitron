**Ans to the question no** 2

```python
inp = input()
numbers = inp.split()
x = float(numbers[0])
y = float(numbers[1])

print("x:",x ,"y:",y)

def compute_ratio(x, y):
  try:
    return (x + y) / (x * y)
  except ZeroDivisionError:
    return "Error: Division not possible"


print(compute_ratio(x, y))
```

**Ans to the question no 3**


```python
import string

with open("./sample_data/data.txt", "r") as file:
  content = file.read()
print(content)

words = content.split()
# print(words)

words = list(map(lambda w: w.strip(string.punctuation), words))
# print(words)

words = list(filter(lambda w:len(w)>= 1, words))
# print(words)

longest_word = max(words, key=len)
print("longest word is: ",longest_word)
```

```python
inp = input()
words = inp.split()
result = list(filter(lambda w: w[0].lower() in ['a', 'e', 'i', 'o', 'u'], words))
print(result)
```

```python
inp = input()
word = inp.lower()
print(word[::-1])
if word == word[::-1]:
  print("Palindrome")
else:
  print("Not Palindrome")
```
